# Task-1: Take an 8-bit grayscale image and apply sharpening operation by using Laplacian filter (with and without build in image sharpening function). Show both output and input image side by side in a subplot with a title.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread("/content/drive/MyDrive/cse 3216/gray_img.jpg", cv2.IMREAD_GRAYSCALE)

filter=np.array([[0, -1, 0],
                 [-1, 4, -1],
                 [0, -1, 0]])

ans=img.copy()
ans=np.pad(ans, 1, 'constant', constant_values=0)

for i in range(1, ans.shape[0]-1):
  for j in range(1, ans.shape[1]-1):
    region=ans[i-1:i+2, j-1:j+2]
    filtered_value=np.sum(region*filter)
    ans[i,j]=np.clip(round(filtered_value),0,255)

ans=ans[1:ans.shape[0]-1, 1:ans.shape[1]-1]

img=np.pad(img, 1, 'constant', constant_values=0)
conv2=cv2.filter2D(img, -1, filter)
img=img[1:img.shape[0]-1, 1:img.shape[1]-1]

plt.figure(figsize=(10,10))
plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title('Input Image')

plt.subplot(1,3,2)
plt.imshow(ans, cmap='gray')
plt.title('Sharpened Img (Manual)')

plt.subplot(1,3,3)
plt.imshow(conv2,cmap='gray')
plt.title('Sharpened Img (Built-in)')

plt.show()
```

# Task-2: Take an 8-bit grayscale img and apply sharpenign operation by using composite laplacian filter( with and without built in img sharpening function). Show both input and output img side by side in a subplot with a title.

# Task-03: Take an 8-bit grayscale image and calculate the gradient of image as well as gradient orientation and gradient magnitude.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread("/content/drive/MyDrive/cse 3216/gray_img.jpg", cv2.IMREAD_GRAYSCALE)

sobelx=cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobely=cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

mag, angle=cv2.cartToPolar(sobelx, sobely, angleInDegrees=True)

plt.figure(figsize=(10,10))

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')

plt.subplot(1,3,2)
plt.imshow(mag, cmap='gray')
plt.title('Gradient Magnitude')

plt.subplot(1,3,3)
plt.imshow(angle, cmap='gray')
plt.title('Gradient Orientation')

plt.show()
```
