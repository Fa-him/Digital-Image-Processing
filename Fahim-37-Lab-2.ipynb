1. Take a 5x6 NumPy array and flip the 5x6 NumPy array using loop.

```python
import numpy as np

original_array=np.array(
    [
        [255, 0, 1, 100, 40, 23],
        [10, 200, 70, 35, 90, 234],
        [190, 23, 44, 53, 56, 80],
        [32, 43, 220, 43, 123, 120],
        [18, 7, 221, 21, 8, 15]
    ]
)

rows, cols=original_array.shape
height=rows
width=cols

flipped_array=original_array.copy()

for i in range(height):
  for j in range(width):
    flipped_array[i][j]=original_array[i][width-j-1]

#for i in range(height):
# flipped_array[i, :]=original_array[i, ::-1]

print('Original Array: ')
print(original_array)
print('\nFlipped Array: ')
print(flipped_array)
```

2. Take a grayscale image and generate the mirror image of the original image. Show input and output image side by side in a subplot with a title.

```python
from google.colab import drive
drive.mount('/content/drive')
import cv2
import matplotlib.pyplot as plt

img=cv2.imread("/content/drive/MyDrive/cse 3216/gray_img.jpg")
img_gray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

height=img_gray.shape[0]
width=img_gray.shape[1]

img_mirrored=np.zeros((height, width))

for i in range(height):
  for j in range(width):
    img_mirrored[i][j]=img_gray[i][width-j-1]

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title('Original img')

plt.subplot(1,2,2)
plt.imshow(img_mirrored, cmap='gray')
plt.title('Mirrored Image')
```

3. take a grayscale image and generate the flipped image of the original image using loop. Show input and output image side by side in a subplot with a title.

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
img=cv2.imread('/content/drive/MyDrive/cse 3216/gray_img.jpg')
img_gray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
height=img_gray.shape[0]
width=img_gray.shape[1]
img_flip=np.zeros((height, width))

for i in range(height):
  for j in range(width):
    img_flip[i][j]=img_gray[height-i-1][j]

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title('Original img')
plt.subplot(1,2,2)
plt.imshow(img_flip, cmap='gray')
plt.title('Flipped img')
```

4. Take an 8-bit grayscale image and apply Thresholding such that the pixel having intensity values below 150 will be replaced by 0 and pixels having intensity value equal or above 150 will be replaced by 255. Display input and output image side by side in a subplot with a title.

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
img=cv2.imread('/content/drive/MyDrive/cse 3216/gray_img.jpg')
img_gray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
height=img_gray.shape[0]
width=img_gray.shape[1]

img_threshold=img_gray.copy()

for i in range(height):
  for j in range(width):
    if img_threshold[i][j]<150:
      img_threshold[i][j]=0
    else:
      img_threshold[i][j]=255

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title('Original img')
plt.subplot(1,2,2)
plt.imshow(img_threshold, cmap='gray')
plt.title('Thresholded img')
```
