3. Print the sum of first 100 numbers using for loop, while loop and do while loop

```python
#using for loop
sum=0
for i in range(1, 101):
  sum+=i
print(sum)

#using while loop
sum=0
i=1
while i<101:
  sum+=i
  i+=1

print(sum)

#using do while
sum=0
i=1
while True:
  sum+=i
  i+=1
  if i>100:
    break

print(sum)
```

4. Print the Fibonacci series using for loop

```python
def print_fibonacchi(n):
  a, b=0,1
  print("Fibonacchi Series:")
  for _ in range(n):
    print(a, end=" ")
    a,b = b, a+b



nums=10
print_fibonacchi(nums)
```

7. Determine the grade of a student using if..else logic if subject number is taken as userinput

```python
def calculate_grade():
  n=int(input("Enter the number of subjects: "))
  totalMarks=0

  for i in range(n):
    marks=float(input(f"Enter marks for the subject {i+1}: "))
    totalMarks+=marks

  averageMarks=totalMarks/n

  if averageMarks>=90:
    grade="A"
  elif averageMarks>=75:
    grade="B"
  elif averageMarks>=60:
    grade="C"
  elif averageMarks>=50:
    grade="D"
  else:
    grade="F"

  print(f"Average marks: {averageMarks}")
  print(f"Grade: {grade}")

calculate_grade()
```

8. Determine the grade of a student using switch..case logic if subject number is taken as user input.

```python
def get_grade(average):
  grade={
      (90, 100):'A+',
      (80, 89):'A',
      (70, 79):'B+',
      (60, 69):'B',
      (50, 59):'C+',
      (40, 49):'C',
      (30, 39):'D',
      (0, 29):'F'
  }

  for tup, i in grade.items():
    if tup[0]<=average<=tup[1]:
      return i
  return "Invalid Marks"

def main():
  n=int(input("Enter the number of subjects: "))
  total_marks=0
  for i in range(n):
    marks=float(input(f"Enter marks for subjects {i+1}: "))
    total_marks+=marks

  average=total_marks/n

  print(f"The avg marks : {average}")

  grade=get_grade(average)
  print(f"Grade is : {grade}")

main()
```
