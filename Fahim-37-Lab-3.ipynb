 Task-01: Take a 5x6 NumPy array and apply negative image operations(calculate image bit dynamically). Convert the input and output NumPy array into an image and show them side by side in a subplot with a title.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

array=np.array([
    [255, 0, 1, 100, 25, 101],
    [10, 200, 70, 80, 120, 150],
    [95, 30, 30, 81, 96, 77],
    [87, 89, 220, 250, 100, 10],
    [18, 7, 221, 21, 8, 15]
])

temp=array.copy()
max_pixel=np.max(array) #finding the max pixel value in the arr
max_pixel=(2**(np.ceil(np.log2(max_pixel))))-1  #L-1 = 255

for i in range(0, 5):
  for j in range(0,6):
    #subtract each pixel value from the max value and replace it
    array[i][j]=max_pixel - array[i][j]

#plot

plt.figure(figsize=(10, 10))
plt.subplot(1,2,1)
plt.title('Original Image')
plt.imshow(temp, cmap='gray')

plt.subplot(1,2,2)
plt.imshow(array, cmap='gray')
plt.title('Negative Image')

```
